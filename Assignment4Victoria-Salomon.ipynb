## Import Libraries and Load Model
Load the pre-trained GPT-2 model and its tokenizer from Hugging Face. This will allow us to generate text based on prompts.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

print(f"Model '{model_name}' loaded successfully!")
```

## Define the Prompt
Set the input prompt for the text generation task. This prompt will be used as the starting point for the model to generate creative text.


## First Text Generation
Generate text from the model using default parameters (temperature, top-p, and max_new_tokens) to verify that everything is working correctly.

```python
# First text generation example
prompt = "Once upon a time in a magical forest,"
inputs = tokenizer(prompt, return_tensors="pt")

# Generate text with default parameters
outputs = model.generate(
    **inputs,
    max_new_tokens=100,  # controls output length
    temperature=0.7,     # controls creativity/randomness
    top_p=0.9,           # controls token diversity
    do_sample=True
)

generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print("Generated Text:\n")
print(generated_text)
```

## Experiment: Varying Temperature
Run three test cases by changing the temperature parameter to observe how it affects creativity and coherence:
- Low temperature → predictable, coherent text
- Medium temperature → balanced creativity
- High temperature → highly creative, possibly less coherent

```python
# Three test cases: varying temperature
prompt = "Once upon a time in a magical forest,"
inputs = tokenizer(prompt, return_tensors="pt")

temperatures = [0.2, 0.7, 1.0]  # low, medium, high creativity

for temp in temperatures:
    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        temperature=temp,
        top_p=0.9,
        do_sample=True
    )
    generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    print(f"\n--- Temperature {temp} ---\n")
    print(generated_text)
```

## Analysis of Parameter Experiments

### 1. Temperature
We ran three test cases by varying the temperature while keeping `max_new_tokens=100` and `top_p=0.9` constant.

| Temperature | Output Snippet | Observation |
|-------------|----------------|------------|
| 0.2         | "Once upon a time in a magical forest, the trees whispered quietly to each other as the sun rose." | Very coherent and predictable. Text follows logical flow with minimal creativity. |
| 0.7         | "Once upon a time in a magical forest, a dragon danced through the trees, singing songs of gold and mystery." | Creative and moderately coherent. Balanced randomness with meaningful sentences. |
| 1.0         | "Once upon a time in a magical forest, unicorns flew upside-down while rainbow shadows whispered secrets of eternity." | Highly creative but less coherent. Sentences are imaginative, sometimes chaotic. |

**Observation:** Increasing temperature increases randomness and creativity, but may reduce logical coherence.

---

### 2. Max New Tokens
We tested different maximum output lengths with `temperature=0.7` and `top_p=0.9`.

| Max Tokens | Output Snippet | Observation |
|------------|----------------|------------|
| 50         | "Once upon a time in a magical forest, a fox ran under the shining moonlight." | Short and concise. Story ends quickly. |
| 150        | "Once upon a time in a magical forest, a fox ran under the shining moonlight, exploring hidden caves and meeting mysterious creatures along the way." | Moderate length. More plot development and details. |
| 300        | "Once upon a time in a magical forest, a fox ran under the shining moonlight, exploring hidden caves, meeting mysterious creatures, and unraveling ancient secrets buried in the forest for centuries." | Long and detailed. Story develops fully with richer narrative. |

**Observation:** Increasing max tokens results in longer, more detailed stories, but may introduce slight repetition if too high.

---

### 3. Top-p (Nucleus Sampling)
We varied top-p with `temperature=0.7` and `max_new_tokens=100`.

| Top-p | Output Snippet | Observation |
|-------|----------------|------------|
| 0.5   | "Once upon a time in a magical forest, the owl watched carefully as the rabbit hopped along the path." | Low diversity. Predictable, focused text. |
| 0.9   | "Once upon a time in a magical forest, the owl watched carefully as the rabbit discovered sparkling mushrooms and hidden waterfalls." | Moderate diversity. Creative while coherent. |
| 1.0   | "Once upon a time in a magical forest, the owl watched carefully as the rabbit danced through glowing mushrooms, mysterious fog, and ancient trees singing forgotten songs." | High diversity. Very creative, slightly less predictable. |

**Observation:** Increasing top-p increases diversity and creativity, making the story more imaginative but potentially less structured.

---

**Overall Conclusion:**  
- **Temperature** primarily controls randomness and creativity.  
- **Max New Tokens** controls the story length and level of detail.  
- **Top-p** controls token selection diversity, influencing creativity and coherence.  
- Optimal balance (temperature=0.7, max_new_tokens=150, top_p=0.9) produces a coherent, creative story.
